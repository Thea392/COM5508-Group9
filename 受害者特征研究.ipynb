```python
import pandas as pd
import time
from zhipuai import ZhipuAI

# ========== 你的配置 ==========
API_KEY = "aec8c161eae440bf9a03fb79803594c0.7uGk8vDg8uOIhPrU"
FILE_PATH = "受害者特征.xlsx"  # 改成你的文件名
OUT_FILE = "经济状况维度_LLM分析.xlsx"
# ================================

client = ZhipuAI(api_key=API_KEY)
df = pd.read_excel(FILE_PATH)

def llm_analyze_economy(text):
    """
    纯 LLM 分析经济状况
    严格返回：
    - 经济水平类型
    
    """
    if pd.isna(text):
        return "无数据", "无数据"

    prompt = f"""
你是专业反诈数据分析研究员。
请根据以下内容，严格分析人物的经济状况：

内容：{text}

请按要求输出这一项：

1. 经济水平类型（只能四选一）：
- 稳定收入
- 低收入
- 高储蓄
- 负债人群

输出格式（固定）：
经济类型：XXX
"""

    try:
        res = client.chat.completions.create(
            model="glm-4-flash",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1
        )
        content = res.choices[0].message.content.strip()

        eco_type = "无数据"
        eco_stress = "无数据"

        for line in content.split("\n"):
            if "经济类型：" in line:
                eco_type = line.replace("经济类型：", "").strip()
            if "经济压力：" in line:
                eco_stress = line.replace("经济压力：", "").strip()

        return eco_type, eco_stress

    except:
        return "分析失败", "分析失败"

# ========== 批量执行 ==========
print("正在使用 LLM 分析经济状况维度...")

eco_types = []
eco_stresses = []

for idx, row in df.iterrows():
    content = str(row["受害者经济状况"]).strip()
    t1, t2 = llm_analyze_economy(content)
    eco_types.append(t1)
    eco_stresses.append(t2)
    time.sleep(1)  # 避免超限

df["经济水平类型_LLM"] = eco_types
df["经济压力大小_LLM"] = eco_stresses

# ========== 过滤无效数据 ==========
df_clean = df[
    (df["经济水平类型_LLM"] != "无数据") &
    (df["经济水平类型_LLM"] != "分析失败") 

].copy()

# ========== 输出统计结果 ==========
print("\n========================================")
print("          经济水平人群受害统计           ")
print("========================================")
eco_count = df_clean["经济水平类型_LLM"].value_counts()
for k, v in eco_count.items():
    print(f"{k}：{v} 人")



# ========== 保存最终文件 ==========
df.to_excel(OUT_FILE, index=False)
print(f"\n✅ 分析完成！文件已保存：{OUT_FILE}")
```

```python
import pandas as pd
import time
from zhipuai import ZhipuAI

# ===================== 你的配置（和之前完全一样）=====================
API_KEY = "aec8c161eae440bf9a03fb79803594c0.7uGk8vDg8uOIhPrU"
FILE_PATH = "受害者特征.xlsx"
OUTPUT_FILE = "综合受害者画像_LLM最终版.xlsx"
MODEL_NAME = "glm-4-flash"
# ==================================================================

client = ZhipuAI(api_key=API_KEY)

# ===================== LLM 综合画像生成 =====================
def generate_comprehensive_profile(row):
    """
    从你已有列读取特征 → 让LLM汇总成一段专业画像
    """
    # 从你的Excel中读取已有的特征
    age = str(row.get("受害者年龄", "未知"))
    job = str(row.get("受害者职业", "未知"))
    finance = str(row.get("受害者经济状况", "未知"))
    motivation = str(row.get("动机倾向", "未知"))
    emotion = str(row.get("情感需求", "未知"))
    stress = str(row.get("压力状态", "未知"))
    loss_aversion = str(row.get("损失厌恶心理", "未知"))
    trust = str(row.get("信任原因", "未知"))

    prompt = f"""
你是专业反诈与犯罪心理分析师。
根据以下受害者的所有特征，生成一段精准、完整、专业的【综合受害者画像】。
语言简洁、学术、可直接用于研究报告。

受害者特征：
年龄：{age}
职业：{job}
经济状况：{finance}
动机倾向：{motivation}
情感需求：{emotion}
压力状态：{stress}
损失厌恶心理：{loss_aversion}
信任原因：{trust}

请输出一段完整的【综合受害者画像】，不要分点、不要多余格式。
"""
    try:
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1
        )
        return response.choices[0].message.content.strip()
    except:
        return "生成失败"

# ===================== 主程序 =====================
if __name__ == "__main__":
    df = pd.read_excel(FILE_PATH)

    print("正在用 LLM 生成综合受害者画像...")

    # 逐行生成综合画像
    df["综合受害者画像_LLM"] = df.apply(generate_comprehensive_profile, axis=1)

    # 保存
    df.to_excel(OUTPUT_FILE, index=False)
    print(f"\n✅ 完成！综合画像已保存到：{OUTPUT_FILE}")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# ===================== 配置 =====================
FILE_PATH = "综合受害者画像_LLM最终版.xlsx"
N_TOPICS = 5  # 想分几类就改几类（3~5类最合适）
# ====================================================

# 1. 读取数据
df = pd.read_excel(FILE_PATH)
df_clean = df[df["诈骗类型"].notna()].copy()

# 2. 提取文本
texts = df_clean["诈骗类型"].astype(str).tolist()

# 3. 文本向量化
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# 4. 训练 LDA 模型（正确用法）
lda = LatentDirichletAllocation(n_components=N_TOPICS, random_state=42)
lda.fit(X)  # 这里只用 fit，不是 fit_predict

# 5. 预测每个文本属于哪个主题（正确写法！）
topic_distribution = lda.transform(X)
df_clean["LDA_聚类主题"] = [f"诈骗主题_{i+1}" for i in topic_distribution.argmax(axis=1)]

# 6. 把结果写回原 Excel（新增一列，不破坏原数据）
df.loc[df_clean.index, "LDA_聚类主题"] = df_clean["LDA_聚类主题"]
df.to_excel(FILE_PATH, index=False)

# ===================== 输出结果 =====================
print("✅ LDA 聚类完成！")
print(f"✅ 已在 Excel 新增一列：LDA_聚类主题")

print("\n" + "="*50)
print("📊 LDA 聚类结果预览")
print("="*50)

for i in range(N_TOPICS):
    theme = f"诈骗主题_{i+1}"
    group = df_clean[df_clean["LDA_聚类主题"] == theme]
    types = group["诈骗类型"].unique()
    print(f"\n{theme}")
    print(" 包含：" + "、".join(types[:5]))
```
